```python
# Ensure 'Datetime' column is parsed as datetime objects using correct capitalization
df['Datetime'] = pd.to_datetime(df['Datetime'])

# 1. Date & Time Extractions
df['year'] = df['Datetime'].dt.year
df['month'] = df['Datetime'].dt.month
df['day'] = df['Datetime'].dt.day
df['hour'] = df['Datetime'].dt.hour
df['date'] = df['Datetime'].dt.date
df['dayoftheweek'] = df['Datetime'].dt.day_name()

# 2. Temperature Group (CASE WHEN)
temp_conds = [
    df['Temperature'] < 10,
    df['Temperature'] < 20,
    df['Temperature'] < 30,
    df['Temperature'] < 40,
]
temp_choices = ['below 10', 'below 20', 'below 30', 'below 40']
df['temperaturegroup'] = np.select(
    temp_conds, temp_choices, default='below 50'
)

# 3. Humidity Group
hum_conds = [
    df['Humidity'] < 20,
    df['Humidity'] < 30,
    df['Humidity'] < 40,
    df['Humidity'] < 50,
    df['Humidity'] < 60,
    df['Humidity'] < 70,
    df['Humidity'] < 80,
    df['Humidity'] < 90,
]
hum_choices = [
    'below 20',
    'below 30',
    'below 40',
    'below 50',
    'below 60',
    'below 70',
    'below 80',
    'below 90',
]
df['humiditygroup'] = np.select(hum_conds, hum_choices, default='below 100')

# 4. Windspeed Group
wind_conds = [
    df['WindSpeed'] < 1,
    df['WindSpeed'] < 2,
    df['WindSpeed'] < 3,
    df['WindSpeed'] < 4,
    df['WindSpeed'] < 5,
    df['WindSpeed'] < 6,
]
wind_choices = [
    'below 1',
    'below 2',
    'below 3',
    'below 4',
    'below 5',
    'below 6',
]
df['windspeedgroup'] = np.select(wind_conds, wind_choices, default='below 7')

# 5. General Diffuse Flows Group
gdf_conds = [
    df['GeneralDiffuseFlows'] < 100,
    df['GeneralDiffuseFlows'] < 200,
    df['GeneralDiffuseFlows'] < 300,
    df['GeneralDiffuseFlows'] < 400,
    df['GeneralDiffuseFlows'] < 500,
    df['GeneralDiffuseFlows'] < 600,
    df['GeneralDiffuseFlows'] < 700,
    df['GeneralDiffuseFlows'] < 800,
    df['GeneralDiffuseFlows'] < 900,
    df['GeneralDiffuseFlows'] < 1000,
    df['GeneralDiffuseFlows'] < 1100,
]
gdf_choices = [
    'below 100',
    'below 200',
    'below 300',
    'below 500',  # matches SQL behavior
    'below 500',
    'below 600',
    'below 700',
    'below 800',
    'below 900',
    'below 1000',
    'below 1100',
]
df['generaldiffuseflowsgroup'] = np.select(
    gdf_conds, gdf_choices, default='below 1200'
)

# 6. Diffuse Flows Group
df_conds = [
    df['DiffuseFlows'] < 100,
    df['DiffuseFlows'] < 200,
    df['DiffuseFlows'] < 300,
    df['DiffuseFlows'] < 400,
    df['DiffuseFlows'] < 500,
    df['DiffuseFlows'] < 600,
    df['DiffuseFlows'] < 700,
    df['DiffuseFlows'] < 800,
    df['DiffuseFlows'] < 900,
]
df_choices = [
    'below 100',
    'below 200',
    'below 300',
    'below 400',
    'below 500',
    'below 600',
    'below 700',
    'below 800',
    'below 900',
]
df['diffuseflowsgroup'] = np.select(df_conds, df_choices, default='below 1000')

# 7. Power Consumption Zone 1 Group
z1_conds = [
    df['PowerConsumption_Zone1'] < 20000,
    df['PowerConsumption_Zone1'] < 30000,
    df['PowerConsumption_Zone1'] < 40000,
    df['PowerConsumption_Zone1'] < 50000,
]
z1_choices = ['below 20000', 'below 30000', 'below 40000', 'below 50000']
df['powerconsumption_zone1_group'] = np.select(
    z1_conds, z1_choices, default='below 60000'
)

# 8. Power Consumption Zone 2 Group
z2_conds = [
    df['PowerConsumption_Zone2'] < 10000,
    df['PowerConsumption_Zone2'] < 20000,
    df['PowerConsumption_Zone2'] < 30000,
]
z2_choices = ['below 10000', 'below 20000', 'below 30000']
df['powerconsumption_zone2_group'] = np.select(
    z2_conds, z2_choices, default='below 40000'
)

# 9. Power Consumption Zone 3 Group
z3_conds = [
    df['PowerConsumption_Zone3'] < 10000,
    df['PowerConsumption_Zone3'] < 20000,
    df['PowerConsumption_Zone3'] < 30000,
    df['PowerConsumption_Zone3'] < 40000,
]
z3_choices = ['below 10000', 'below 20000', 'below 30000', 'below 40000']
df['powerconsumption_zone3_group'] = np.select(
    z3_conds, z3_choices, default='below 50000'
)

display(df.head())
```

```python
import pandas as pd
pd.plotting.register_matplotlib_converters()
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
print("Setup Complete")
```

```python
plt.figure(figsize=(16,10))
plt.title('Power Consumption over time')
sns.lineplot(data=df, x="date", y="PowerConsumption_Zone1")
plt.show()
```

```python
plt.figure(figsize=(16,10))
plt.title('Power Consumption over time')
sns.lineplot(data=df, x="date", y="PowerConsumption_Zone2")
plt.show()
```

```python
plt.figure(figsize=(16,10))
plt.title('Power Consumption over time')
sns.lineplot(data=df, x="date", y="PowerConsumption_Zone3")
plt.show()
```

```python
# Calculate average consumption for each zone
avg_consumption = df[['PowerConsumption_Zone1', 'PowerConsumption_Zone2', 'PowerConsumption_Zone3']].mean()

# Create a bar plot
plt.figure(figsize=(10, 6))
sns.barplot(x=avg_consumption.index, y=avg_consumption.values, palette='viridis', hue=avg_consumption.index, legend=False)
plt.title('Average Power Consumption by Zone')
plt.ylabel('Average Consumption (Units)')
plt.xlabel('Zones')
plt.show()
```

```python
# Group by hour and calculate mean consumption for each zone
hourly_avg = df.groupby('hour')[['PowerConsumption_Zone1', 'PowerConsumption_Zone2', 'PowerConsumption_Zone3']].mean()

# Plot hourly patterns
plt.figure(figsize=(14, 8))
plt.plot(hourly_avg.index, hourly_avg['PowerConsumption_Zone1'], label='Zone 1', marker='o', linewidth=2)
plt.plot(hourly_avg.index, hourly_avg['PowerConsumption_Zone2'], label='Zone 2', marker='s', linewidth=2)
plt.plot(hourly_avg.index, hourly_avg['PowerConsumption_Zone3'], label='Zone 3', marker='^', linewidth=2)

plt.title('Hourly Power Consumption Patterns by Zone', fontsize=14)
plt.xlabel('Hour of Day (24-Hour Clock)', fontsize=12)
plt.ylabel('Average Consumption (Units)', fontsize=12)
plt.xticks(range(0, 24))
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(fontsize=11)
plt.show()
```

```python
# Group by month and calculate mean consumption for each zone
monthly_avg = df.groupby('month')[['PowerConsumption_Zone1', 'PowerConsumption_Zone2', 'PowerConsumption_Zone3']].mean()

# Plot monthly patterns
plt.figure(figsize=(14, 8))
plt.plot(monthly_avg.index, monthly_avg['PowerConsumption_Zone1'], label='Zone 1', marker='o', linewidth=2)
plt.plot(monthly_avg.index, monthly_avg['PowerConsumption_Zone2'], label='Zone 2', marker='s', linewidth=2)
plt.plot(monthly_avg.index, monthly_avg['PowerConsumption_Zone3'], label='Zone 3', marker='^', linewidth=2)

plt.title('Monthly Power Consumption Patterns by Zone', fontsize=14)
plt.xlabel('Month of Year', fontsize=12)
plt.ylabel('Average Consumption (Units)', fontsize=12)
plt.xticks(range(1, 13), ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(fontsize=11)
plt.show()
```

```python
# Define the logical order of days for sorting
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

# Group by day of week and calculate mean consumption for each zone, then reindex to match logical order
weekly_avg = df.groupby('dayoftheweek')[['PowerConsumption_Zone1', 'PowerConsumption_Zone2', 'PowerConsumption_Zone3']].mean().reindex(day_order)

# Plot day-of-week patterns
plt.figure(figsize=(14, 8))
plt.plot(weekly_avg.index, weekly_avg['PowerConsumption_Zone1'], label='Zone 1', marker='o', linewidth=2)
plt.plot(weekly_avg.index, weekly_avg['PowerConsumption_Zone2'], label='Zone 2', marker='s', linewidth=2)
plt.plot(weekly_avg.index, weekly_avg['PowerConsumption_Zone3'], label='Zone 3', marker='^', linewidth=2)

plt.title('Average Power Consumption by Day of Week', fontsize=14)
plt.xlabel('Day of the Week', fontsize=12)
plt.ylabel('Average Consumption (Units)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(fontsize=11)
plt.show()
```

```python
# Select environmental factors and power consumption columns
env_cols = ['Temperature', 'Humidity', 'WindSpeed', 'PowerConsumption_Zone1', 'PowerConsumption_Zone2', 'PowerConsumption_Zone3']
corr_matrix = df[env_cols].corr()

# Plot correlation heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5, vmin=-1, vmax=1)
plt.title('Correlation between Environmental Factors and Power Consumption', fontsize=14)
plt.show()
```

```python
plt.figure(figsize=(16, 10))
plt.title('Comparison of Power Consumption Across All Three Zones', fontsize=14)
sns.lineplot(data=df, x='date', y='PowerConsumption_Zone1', label='Zone 1')
sns.lineplot(data=df, x='date', y='PowerConsumption_Zone2', label='Zone 2')
sns.lineplot(data=df, x='date', y='PowerConsumption_Zone3', label='Zone 3')
plt.ylabel('Power Consumption (Units)', fontsize=12)
plt.xlabel('Date', fontsize=12)
plt.legend(fontsize=11)
plt.show()
```
